Update `RUN_DIR` below after you generate evaluation artifacts.

```python
from pathlib import Path
import json
import matplotlib.pyplot as plt
import pandas as pd
from PIL import Image

RUN_DIR = Path('../runs/heart_unet/eval')
summary = json.loads((RUN_DIR / 'evaluation_summary.json').read_text())
metrics = pd.read_csv(RUN_DIR / 'case_metrics.csv')
summary
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
metrics['dice'].plot(kind='hist', bins=10, ax=axes[0], title='Dice')
metrics['inference_seconds'].plot(kind='hist', bins=10, ax=axes[1], title='Inference Seconds')
plt.tight_layout()
```

```python
figure_paths = sorted((RUN_DIR / 'figures').glob('*.png'))[:5]
for figure_path in figure_paths:
    plt.figure(figsize=(12, 4))
    plt.imshow(Image.open(figure_path))
    plt.title(figure_path.stem)
    plt.axis('off')
    plt.show()
```
